```python
import pandas as pd
import numpy as np
import os


# PATH DATASET

base_path = "/content/drive/My Drive/Colab Notebooks/DATASET/datasetgabung/"

files = [
    "jalanjatuh.csv",
    "diamjatuh.csv",
    "jalan.csv",
    "diam.csv"
]


# OUTPUT FOLDER

output_acc = os.path.join(base_path, "accelero_only")
output_gyro = os.path.join(base_path, "gyro_only")

os.makedirs(output_acc, exist_ok=True)
os.makedirs(output_gyro, exist_ok=True)

# PROCESS DATASET

for file in files:

    file_path = os.path.join(base_path, file)

    print(f"MEMPROSES : {file}")

    # BACA CSV


    df = pd.read_csv(file_path)


    # CEK NAMA KOLOM


    print("KOLOM :", df.columns.tolist())

    # ACCELEROMETER ONLY

    df_acc = df[[
        "ax",
        "ay",
        "az"
    ]]

    acc_path = os.path.join(
        output_acc,
        file
    )

    df_acc.to_csv(
        acc_path,
        index=False
    )

    # GYROSCOPE ONLY

    df_gyro = df[[
        "gx",
        "gy",
        "gz"
    ]]

    gyro_path = os.path.join(
        output_gyro,
        file
    )

    df_gyro.to_csv(
        gyro_path,
        index=False
    )

    print(f"ACC SAVE : {acc_path}")
    print(f"GYRO SAVE : {gyro_path}")

print("\nSELESAI MEMBUAT DATASET")
```

```python
#resultan_acc_only

import pandas as pd
import numpy as np
import os

# PATH ACCELERO ONLY


base_path = "/content/drive/My Drive/Colab Notebooks/DATASET/datasetgabung/accelero_only/"

files = [
    "jalanjatuh.csv",
    "diamjatuh.csv",
    "jalan.csv",
    "diam.csv"
]

# OUTPUT

output_path = os.path.join(
    base_path,
    "acc_resultant"
)

os.makedirs(
    output_path,
    exist_ok=True
)

# PROCESS

for file in files:

    file_path = os.path.join(
        base_path,
        file
    )

    print(f"MEMPROSES : {file}")

    df = pd.read_csv(file_path)

    # RESULTAN ACCELEROMETER

    df["acc_resultant"] = np.sqrt(

        (df["ax"] ** 2) +
        (df["ay"] ** 2) +
        (df["az"] ** 2)

    )

    # AMBIL RESULTAN SAJA

    df_result = df[[

        "acc_resultant"

    ]]

    # SAVE

    save_path = os.path.join(
        output_path,
        file
    )

    df_result.to_csv(
        save_path,
        index=False
    )

    print(f"SAVED : {save_path}")

print("SELESAI")
```

```python
# PSA GYRO ONLY
# POWER SPECTRUM ANALYSIS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.signal import welch

fs = 200

path = "/content/drive/My Drive/Colab Notebooks/DATASET/datasetgabung/gyro_only/"

kelas_list = [
    "diam",
    "jalan",
    "diamjatuh",
    "jalanjatuh"
]

channels = [
    "gx",
    "gy",
    "gz"
]

for kelas in kelas_list:

    file = path + kelas + ".csv"

    df = pd.read_csv(file)

    print("KELAS :", kelas.upper())
    print("Jumlah Data :", len(df))

    fig, axs = plt.subplots(
        3,
        1,
        figsize=(14,10)
    )

    for i, ch in enumerate(channels):

        signal = df[ch].values

        freq, psd = welch(
            signal,
            fs=fs,
            nperseg=2048
        )

        idx = np.argmax(psd[1:]) + 1

        dom_freq = freq[idx]
        dom_power = psd[idx]

        print(
            f"{ch} -> Dominant Freq = "
            f"{dom_freq:.3f} Hz | "
            f"Power = {dom_power:.5f}"
        )

        axs[i].plot(
            freq,
            psd,
            linewidth=1.5
        )

        axs[i].set_title(
            f"{kelas} - {ch}"
        )

        axs[i].set_xlabel("Frequency (Hz)")

        axs[i].set_ylabel("Power")

        axs[i].set_xlim(0,20)

        axs[i].grid(True)

    plt.tight_layout()

    plt.show()
```

```python
#PSA_acc_only
#power_spectrum_analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

base_path = "/content/drive/My Drive/Colab Notebooks/DATASET/datasetgabung/accelero_only/"

files = {
    "diam.csv": 0,
    "jalan.csv": 1,
    "diamjatuh.csv": 2,
    "jalanjatuh.csv": 3
}

all_data = []

for file, label in files.items():

    file_path = os.path.join(
        base_path,
        file
    )

    df = pd.read_csv(file_path)

    df = df[[
        "ax",
        "ay",
        "az"
    ]]

    df["label"] = label

    all_data.append(df)

dataset = pd.concat(
    all_data,
    ignore_index=True
)

X = dataset[[
    "ax",
    "ay",
    "az"
]]

y = dataset["label"]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10,8))

scatter = plt.scatter(

    X_pca[:,0],
    X_pca[:,1],

    c=y,

    alpha=0.5
)

plt.xlabel("PCA 1")

plt.ylabel("PCA 2")_

plt.title("PCA Accelerometer Only")

plt.colorbar(scatter)

plt.grid(True)

plt.show()
```

```python
# PCA ACC + GYRO RESULTANT

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

base_path = "/content/drive/My Drive/Colab Notebooks/DATASET/datasetgabung/accgyro_resultant/"

files = {
    "diam.csv": 0,
    "jalan.csv": 1,
    "diamjatuh.csv": 2,
    "jalanjatuh.csv": 3
}

all_data = []

for file, label in files.items():

    file_path = os.path.join(
        base_path,
        file
    )

    df = pd.read_csv(file_path)

    df = df[[
        "acc_resultant",
        "gyro_resultant"
    ]]

    df["label"] = label

    all_data.append(df)

dataset = pd.concat(
    all_data,
    ignore_index=True
)

X = dataset[[
    "acc_resultant",
    "gyro_resultant"
]]

y = dataset["label"]

# STANDARDIZATION

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# VISUALIZATION

plt.figure(figsize=(10,8))

scatter = plt.scatter(

    X_pca[:,0],
    X_pca[:,1],

    c=y,

    alpha=0.5
)

plt.xlabel("PCA 1")

plt.ylabel("PCA 2")

plt.title("PCA Acc + Gyro Resultant")

plt.colorbar(scatter)

plt.grid(True)

plt.show()

# EXPLAINED VARIANCE

print("Explained Variance Ratio :")
print(pca.explained_variance_ratio_)
```
